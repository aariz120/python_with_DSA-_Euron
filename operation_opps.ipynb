## Abstraction in Oops

Abstraction is a fundamental concept in python programming that allow us to simplify complex concepts and focus on the essential details.It involves hiding unneessary details and exposing only the relevant information to the users

```python
# Abstract base Class (ABC)
from abc import ABC,abstractmethod


# senior 

class bankServer(ABC):
    def database(self):
        print("Connected to database")
        
        
    @abstractmethod
    def security(self):
        print("security passed ")
        
        
        
#junior
class MobileApp(bankServer):
    def mobile_login(self):
        print("Mpbile app login successful")
        
        
    def security(self):
        print("Mobile app security")
```

```python
mob=MobileApp()
```

```python
mob.mobile_login()
```

```python
mob.database()
```

```python
mob.security()
```
