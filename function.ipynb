## Decorator

```python
def say_hi():
    print("hi euron")
```

```python
say_hi()
```

```python
def decorator_fun(func):
    def wrapper():
        print("you will be printed before function")
        func()
        print("you will be printed after function call")
    return wrapper    
```

```python
@decorator_fun
def say_hi():
    print("hi euron")
```

```python
say_hi()
```

```python
import time
def timer(func):
    def wrapper(*args,**kwargs):
        
        start=time.time()
        result=func(*args,**kwargs)
        print("time taken by this function is ",time.time()-start)
        return result
    return wrapper 
    
```

```python
@timer
def listparser(l):
    time.sleep(1)
    print(l)
    for i in l:
        print(i)
```

```python
listparser([4,5,56,6,7,8,9,8])
```

```python
def decorator_fun1(func):
    def wrapper(*args,**kwargs):
        print("you will be printed before function")
        func(*args,**kwargs)
        print("you will be printed after function call")
    return wrapper
```

```python
@timer
@decorator_fun1
def listparser(l):
    time.sleep(1)
    print(l)
    for i in l:
        print(i)
```

```python
listparser([4,5,56,6,7,8,9,8])
```
