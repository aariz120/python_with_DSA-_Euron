# break, continue, pass ,else in loop

```python
for i in range (1,20):
    if i%7==0:
        break
    print(i)
```

```python
for i in range(1,20):
    if i %2==0:
        continue
    print(i)
```

```python
for i in range(1,20):
    if i ==3:
        pass
    print(i)
```

```python
while True:
    name=input("enter your name:")
    if name=="eesha":
        break
    print(name)
    
```

```python
i=0
while i <5:
    i=i+1
    if i ==3:
        continue
    print(i)
```

```python
i=0
while i <5:
    i=i+1
    if i ==3:
        pass
    print(i)
```

```python
i=0
while i <5:
    i=i+1
    if i ==3:
        break
    print(i)
```

```python
for i in range(5):
    print(i)
else:
    print("this will be executed itself if for loop is going to complete")
```

```python
for i in range(5):
    if i ==3:
        break
    print(i)
else:
    print("done")
```

```python
x=0
while x<3:
    print(x)
    x=x+1
else:
    print("this will be executed if while loop ends naturally")
```

```python
x=0
while x<3:
    print(x)
    if x==2:
        break
    x=x+1
else:
    print("this will be executed if while loop ends naturally")
```
