```python
day=input(" enter the day :")
match day :
    case "monday":
        print("this is a first day of the week")
    case "tuesday":
        print(" this is second day of the week")
        
    case "wednesday":
        print("this is the third day of the week")
        
    case "thrusday":
        print(" this is the fourth day of the week")
        
    case "friday":
        print(" this is the fifth day of the week")
    case _:
        print(" it's weekend enjoy your day")
        
```
