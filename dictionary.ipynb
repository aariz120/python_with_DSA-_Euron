## List , Set and Dictionary comprehension

```python
range(1,6)
```

```python
l=[]
for i in range(1,6):
    print(i**2)
    l.append(i**2)
  
```

```python
l
```

```python
[i**2 for i in range(1,6)]
```

```python
[i for i in range(1,20) if i%2==0]
```

```python
l=[]
for i in range(20):
    if i%2==0:
        l.append(i)
```

```python
l
```

```python
words=["apple","banana","cherry"]
[i.upper() for i in words]
```

```python
l=[[3,4],[6,7]]
l_final=[3,4,5,6]
```

```python
[i for i in l ]
```

```python
[j for i in l for j in i]
```

```python
[i for i in range(20) if i%2==0 if i%3==0]
```

## Set comprehension

```python
{i for i in [4,5,6,7,87,89,3,4,5,6,45,57,3]}
```

```python
l=["apple","banana","cherry","pineapple","Guava"]
```

```python
{i[0]for i in l}
```

```python
{i:i**2 for i in range(5)}
```

```python
s="eeshu"
```

```python
{i:d for i,d in enumerate(s)}
```
