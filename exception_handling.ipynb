```python
class xyz(Exception):
    pass
```

```python
def register(age):
    if age<0:
        raise xyz("age can't be negative")
```

```python
try:
    register(-67)
except xyz as e:
    print(e)
```

```python
class Toosortpassword(Exception):
    def __init__(self,length):
        self.lenght=length
        super().__init__(f"password is too sort {length} characters found")
```

```python
def validate_password_length(password):
    if len(password)<8:
        raise Toosortpassword(len(password))    
```

```python
validate_password_length("23456789")
```

```python
try:
    validate_password_length("23456")
except Toosortpassword as e:
    print(e)
```
