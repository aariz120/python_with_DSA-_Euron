## CSV file handling

```python
import csv
```

```python
with open('sample.csv','r') as f:
    reader=csv.reader(f)
    print(reader)
    for i in reader:
        print(i)
```

```python
with open('sample.csv','r') as f:
    reader=csv.reader(f)
    next(reader)
    print(reader)
    for i in reader:
        print(i)
```

```python
with open('output.csv','w',newline='') as f:
     writer=csv.writer(f)
     writer.writerow(['name','company'])
     writer.writerow(['name1','company1'])
```

```python
with open('output.csv','w',newline='') as f:
     writer=csv.writer(f)
     writer.writerows([['name','company'],['name1','company1'],['name2','company2'],['name3','company3'],['name4','company4']])
```

```python
with open('output.csv','r',newline='') as f:
     reader=csv.DictReader(f)
     for i in reader:
          print(i)
     
```

## Json File handling

```python
j={
    "employes":[
        {"name":"aariz","age":40},
        {"name":"rakshanda","age":35}
    ]
}
```

```python
import json
```

```python
with open('data.json','r')as f:
    data=json.load(f)
    print(data)
```

```python
with open('data.json','r')as f:
    data=json.load(f)
    for i in data['employes']:
        print(i)
        print(i['name'],i['age'])
```
