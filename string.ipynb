```python
from string import Template
```

```python
Template("my name is $name my age is $age").substitute(name=name, age=age)
```

```python
email_template=Template("""
                        From:$sender
                        To:$receiver
                        Subject:$subject
                        $body
                        """)
```

```python
email_template.substitute(sender="aariz120@gmail.com",receiver="sskumar@gmail.com",
                          subject="meeting postpone ",body="meeting has been postponed to next week")
```

```python
email_template.substitute(sender=input("enter sender information"),receiver="sskumar@gmail.com",
                          subject="meeting postpone ",body="meeting has been postponed to next week")
```
