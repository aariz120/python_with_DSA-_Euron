```python
# MRO (method resolution order)

class a:
    def show(self):
        print("a")
        
class b(a):
    def show(self):
        print("b")
        super().show()
class c(a):
    def show(self):
        print("c")
        super().show()
class d(b,c):
    def show(self):
        print("d")
        super().show()
```

```python
d_obj=d()
d
```

```python
d_obj.show()
```

```python
d.mro()
```

```python
class course(ABC):
    @abstractmethod
    def content(self):
        pass
    @abstractmethod
    def duration(self):
        pass
```

```python
class pythoncourse(course):
    def content(self):
        return "this method will give you content"
    def duration(self):
        return "duration of python course is 3 months"
```
